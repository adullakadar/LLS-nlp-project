```python
nltk_setup()
video = 'Y8yTY_7w3KA'
nlp = spacy_setup()
embedding = embedding_setup()
youtube_api = api_setup()
```

```python
comments_1_df = extract_video_comments_full(video,youtube_api,nlp, embedding)
comments_1_df.head()
```

```python
video2 = 'V0wtNI-rzSg'
comments_2_df = extract_video_comments_full(video2,youtube_api,nlp, embedding)
comments_2_df.head()
```

```python
transcript_1_df = extract_video_transcript_full(video,nlp,embedding,win=60)
transcript_2_df = extract_video_transcript_full(video2,nlp,embedding,win=60)
```

```python
all_dfs = []
all_dfs.append(comments_1_df)
all_dfs.append(comments_2_df)
all_dfs.append(transcript_1_df)
all_dfs.append(transcript_2_df)
meta_df = pd.concat(all_dfs, ignore_index=True)
print(len(meta_df), "total rows")
print(meta_df["source_type"].value_counts())
print(meta_df["video_id"].value_counts())
```

```python
texts, vectors, metadata = load_corpus_from_df(meta_df)
vs = build_faiss(texts, vectors, metadata, embedding)
save_df_csv(meta_df,f'meta_data.csv',True)
```

```python
from modules.retrieval import semantic_search

for q in ["is tesla better than byd?", "what is the battery capacity and range?"]:
    print(f"\nQ: {q}")
    for r in semantic_search(vs, q, k=5):
        print(r["source_type"], "|", r["text"][:70])
```

```python
print("--- transcripts only ---")
for r in filtered_search(vs, "what is the battery capacity and range?",
                         source_type="transcript", k=5):
    print(r["source_type"], "|", r["text"][:70])

print("\n--- comments only ---")
for r in filtered_search(vs, "is tesla better than byd?",
                         source_type="comment", k=5):
    print(r["source_type"], "|", r["text"][:70])
```

```python
agent.VECTORSTORE = vs
graph = build_graph()

with mlflow.start_run(run_name='rag_router_test'):
  for q in ['what is the battery capacity?', 'is tesla better than byd']:
    result = graph.invoke({'query': q})
    print(f'\nQ: {q}')
    print(f'category: {result["category"]}')
    print(f'answer: {result["answer"]}')
```

```python
HARD_QUERIES = [
    {"query": "is the 82.5 kWh battery good?",                      "expected": "opinion"},  # fact + judgment
    {"query": "how does the range compare to what people hoped?",  "expected": "opinion"},  # spec + sentiment
    {"query": "what's the most complained-about feature?",         "expected": "opinion"},  # factual framing, opinion content
    {"query": "do the specs justify the price?",                   "expected": "opinion"},  # specs + value judgment
]
```

```python
acc = evaluate_routing(graph, test_queries=HARD_QUERIES,model_name="qwen3:8b")
```
